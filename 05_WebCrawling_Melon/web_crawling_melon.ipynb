# **멜론 음원 차트 크롤링 (Melon Chart Web Crawling)**

**<분석 목적>**  
Selenium을 활용해 멜론(Melon) 음원 사이트의 인기 차트 상위 100개 곡 정보를 자동으로 수집하는 크롤링 로직 설계

```python
from selenium import webdriver
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.common.by import By
import time
import random
import pandas as pd


url = 'https://www.melon.com/index.htm'
driver = webdriver.Chrome()
driver.get(url)
time.sleep(random.uniform(2, 5))

driver.find_element(By.XPATH, '//*[@id="gnb_menu"]/ul[1]/li[1]/a/span[2]').click()
time.sleep(random.uniform(2, 5))

rank_list= ["lst50", "lst100"]   

data=[]
for i in rank_list:
    el1=driver.find_elements(By.XPATH, f'//*[@id="{i}"]/td[2]/div/span[1]')
    el2=driver.find_elements(By.XPATH, f'//*[@id="{i}"]/td[6]/div/div')
    el3=driver.find_elements(By.XPATH, f'//*[@id="{i}"]/td[7]/div/div')
    el4=driver.find_elements(By.XPATH, f'//*[@id="{i}"]/td[8]/div/button/span[2]')
    
    for i in range(1, len(el1)+1) :
        data.append({
        "순위":el1[i-1].text,
        "곡정보":el2[i-1].text,
        "앨범":el3[i-1].text,
        "좋아요 수":el4[i-1].text})         
df = pd.DataFrame(data)
print(df)
df.to_csv("C:/Users/joyce/OneDrive/바탕 화면/바다 학회/melon_top100.csv", index=False, encoding="utf-8-sig")
time.sleep(3)



```

```python
# =========================================================================
# Step 1. 필수 라이브러리 import
# =========================================================================
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import time
import random
import pandas as pd
from webdriver_manager.chrome import ChromeDriverManager
import logging

# URLLIB3 경고 메시지 레벨 설정
logging.getLogger('urllib3').setLevel(logging.WARNING)

# 크롤링 목표 설정
REVIEW_URL = "https://map.naver.com/p/entry/place/421409079?c=15.00,0,0,0,dh&placePath=/review?additionalHeight=76&fromPanelNum=1&locale=ko&svcName=map_pcv5&timestamp=202511251249&fromPanelNum=1&locale=ko&svcName=map_pcv5&timestamp=202511251249&additionalHeight=76"
TARGET_COUNT = 100 
all_reviews = []

# =========================================================================
# Step 2. 로컬 환경 WebDriver 설정 (버전 명시)
# =========================================================================
print("\nWebDriver 설정 시작...")

# 🚨 사용자 정의: 이 부분에 현재 PC의 Chrome 브라우저 버전을 입력하세요.
CHROME_VERSION = "118.0.5993.88" # [여기에 실제 PC의 Chrome 버전 입력]

# WebDriverManager를 사용하여 명시된 버전의 ChromeDriver를 설치
service = Service(ChromeDriverManager(version=CHROME_VERSION).install()) 

chrome_options = Options()
chrome_options.add_argument('user-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/100.0.4896.75 Safari/537.36')
# chrome_options.add_argument('--headless') # 로컬에서 창을 숨기려면 이 주석을 해제하세요.

try:
    driver = webdriver.Chrome(service=service, options=chrome_options)
    driver.get(REVIEW_URL)
    print("✅ Chrome WebDriver 실행 및 페이지 접속 성공.")
    time.sleep(random.uniform(3, 5)) # 초기 페이지 로딩을 위해 충분히 대기

    # =========================================================================
    # Step 3. 네이버 지도 iframe으로 컨텍스트 전환
    # =========================================================================
    
    # iframe이 로드될 때까지 15초 대기 후 전환
    WebDriverWait(driver, 15).until(
        EC.frame_to_be_available_and_switch_to_it((By.ID, "entryIframe"))
    )
    print("iframe 컨텍스트 전환 성공.")
    
    # 나머지 로직은 이전과 동일하게 진행됩니다. (스크롤 및 리뷰 추출)
    # ... (이하 생략)

    # -------------------------------------------------------------------------
    # Step 3 후속 로직 (스크롤 및 추출)
    # -------------------------------------------------------------------------
    
    WebDriverWait(driver, 10).until(
        EC.presence_of_element_located((By.CSS_SELECTOR, "div.pui_vn15T2 a"))
    )
    
    scroll_container_selector = "div.place_section_content" 
    scroll_count = 0
    while len(all_reviews) < TARGET_COUNT:
        
        try:
            unfolded_links = WebDriverWait(driver, 2).until(
                EC.presence_of_all_elements_located((By.CSS_SELECTOR, "a[data-pui-click-code='rvshowmore']"))
            )
            for link in unfolded_links:
                driver.execute_script("arguments[0].click();", link)
            time.sleep(random.uniform(0.5, 1))
        except:
            pass
            
        review_elements = driver.find_elements(By.CSS_SELECTOR, "div.pui_vn15T2 a")
        current_reviews = []
        for element in review_elements:
            text = element.text.strip()
            if text and len(text) > 5:
                current_reviews.append(text)
        
        newly_collected_count = 0
        for review in current_reviews:
            if review not in all_reviews:
                all_reviews.append(review)
                newly_collected_count += 1
                
        print(f"현재 수집된 리뷰: {len(all_reviews)}개 / 목표: {TARGET_COUNT}개 (새로 {newly_collected_count}개 추가)")

        if len(all_reviews) >= TARGET_COUNT:
            break

        try:
            scroll_element = driver.find_element(By.CSS_SELECTOR, scroll_container_selector)
            driver.execute_script("arguments[0].scrollTop = arguments[0].scrollHeight", scroll_element)
            time.sleep(random.uniform(2, 3))
            scroll_count += 1
        except Exception as e:
            print(f"스크롤 실패: {e}.")
            break
            
        if scroll_count > 50 and newly_collected_count == 0:
            print("무한 스크롤 방지: 크롤링을 종료합니다.")
            break

    # =========================================================================
    # Step 5. 최종 데이터 정리 및 CSV 저장
    # =========================================================================
    
    final_reviews = all_reviews[:TARGET_COUNT]
    
    df = pd.DataFrame(final_reviews, columns=['리뷰_텍스트'])
    csv_filename = '은평점_네이버_리뷰_{}개.csv'.format(len(final_reviews))
    df.to_csv(csv_filename, index=False, encoding='utf-8-sig')
    
    print(f"\n==================================================")
    print(f"✅ 최종 {len(final_reviews)}개의 리뷰를 수집 완료했습니다.")
    print(f"💾 파일명: {csv_filename} (저장 위치: 현재 Jupyter Notebook 폴더)")
    print(f"==================================================")
    
except Exception as e:
    print(f"\n❌ 크롤링 중 최종 오류 발생: {e}")
    
finally:
    driver.quit()
```
